# Search evaluation: which text should we embed?

We compare three text variants on the full dataset (1,590 movies) using the example queries from the assignment plus a few of my own.

```python
import sys
sys.path.append("..")  

import pandas as pd
from data_prep import load_data
from search import TEXT_VARIANTS, build_embedding_text, embed_texts, load_model, search
```

```python
movies, availability, consumption, snapshot = load_data()
model = load_model()
movies.shape
```

## Embeddings for each variant
This takes a little while: 1,590 movies × 3 variants.

```python
embeddings = {}
for variant in TEXT_VARIANTS:
    texts = build_embedding_text(movies, variant)
    embeddings[variant] = embed_texts(model, texts)
    print(variant, embeddings[variant].shape)
```

## Compare top 5 results side by side

```python
def compare(query, top_k=5):
    columns = {}
    for variant in TEXT_VARIANTS:
        res = search(query, model, embeddings[variant], movies, top_k)
        columns[variant] = (
            res["original_title"] + " (" + res["year"].astype(str) + ") "
            + res["score"].round(3).astype(str)
        )
    print(f"Query: {query!r}")
    return pd.DataFrame(columns)
```

### Queries from the assignment
